```python
# First Order Logic - Forward Chaining

# Initial Knowledge Base
KB = {
    ("American", "Robert"),
    ("Owns", "A", "T1"),
    ("Missile", "T1"),
    ("Enemy", "A", "America")
}

# Forward Chaining Algorithm
def forward_chaining(KB, query):

    while True:
        new_facts = set()

        for fact in KB:

            # Rule 1: Missiles are weapons
            if fact[0] == "Missile":
                new_facts.add(("Weapon", fact[1]))

            # Rule 2: Enemies of America are hostile
            if fact[0] == "Enemy" and fact[2] == "America":
                new_facts.add(("Hostile", fact[1]))

        # Rule 3: Robert sells missiles owned by Country A
        for missile in KB:
            if missile[0] == "Missile":
                x = missile[1]

                if ("Owns", "A", x) in KB:
                    new_facts.add(("Sells", "Robert", x, "A"))

        # Rule 4: Americans selling weapons to hostile nations
        # are criminals
        for american in KB:
            if american[0] == "American":
                person = american[1]

                for weapon in KB:
                    if weapon[0] == "Weapon":
                        item = weapon[1]

                        for sale in KB:
                            if sale[0] == "Sells":
                                seller, sold_item, country = sale[1:]

                                if seller == person and sold_item == item:
                                    if ("Hostile", country) in KB:
                                        new_facts.add(
                                            ("Criminal", person)
                                        )

        # Remove already existing facts
        new_facts = new_facts - KB

        # If no new facts are generated, stop
        if not new_facts:
            break

        # Add newly inferred facts
        KB.update(new_facts)

        # Check query
        if query in KB:
            return True

    return query in KB


# Query
query = ("Criminal", "Robert")

# Execute Forward Chaining
if forward_chaining(KB, query):
    print("Query proved: Robert is a criminal.")
else:
    print("Query could not be proved.")
```
